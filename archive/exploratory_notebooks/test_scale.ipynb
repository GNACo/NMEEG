```python
import numpy as np
import matplotlib.pyplot as plt
from mne.time_frequency import psd_array_welch
from scipy.interpolate import interp1d

def corregir_escala_a_uV(raw_dict):
    raw_dict_corregido = {}

    print("Centro       | Max (abs)   | Std         | Escala detectada             | Acción aplicada")
    print("-" * 90)

    for name, raw in raw_dict.items():
        raw.load_data()
        data = raw.get_data()
        max_val = np.abs(data).max()
        std_val = data.std()
        accion = ""
        
        if max_val > 1e4:
            escala = "¿milivoltios mal escalados?"
            raw._data /= 1e3
            accion = "÷ 1e3 (mV → μV)"
        elif max_val > 100:
            escala = "μV"
            accion = "OK"
        elif max_val > 0.01:
            escala = "milivoltios"
            raw._data *= 1e3
            accion = "× 1e3 (mV → μV)"
        elif max_val > 1e-4:
            escala = "voltios"
            raw._data *= 1e6
            accion = "× 1e6 (V → μV)"
        else:
            escala = "muy bajo (posible flat)"
            accion = "⚠️ no modificado"

        print(f"{name:12s} | {max_val:10.3e} | {std_val:10.3e} | {escala:28s} | {accion}")
        raw_dict_corregido[name] = raw

    return raw_dict_corregido

def plot_psd_all_centers_interpolated(raw_dict, fmin=0.5, fmax=45.0, common_freq_res=0.5):
    psds_interp = {}
    common_freqs = np.arange(fmin, fmax + common_freq_res, common_freq_res)

    for site_name, raw in raw_dict.items():
        raw.load_data()
        data = raw.get_data()
        sfreq = raw.info['sfreq']
        n_times = data.shape[1]
        n_fft = min(2048, n_times)

        psd, freqs = psd_array_welch(
            data, sfreq=sfreq, fmin=fmin, fmax=fmax,
            n_fft=n_fft, verbose=False
        )

        psd_mean = psd.mean(axis=0)

        # Interpolación al eje común
        f_interp = interp1d(freqs, psd_mean, kind='linear', bounds_error=False, fill_value="extrapolate")
        psd_interp = f_interp(common_freqs)
        psds_interp[site_name] = psd_interp

    # Plot
    plt.figure(figsize=(12, 6))
    for site_name, psd_vals in psds_interp.items():
        plt.plot(common_freqs, psd_vals, label=site_name)

    plt.xlabel('Frecuencia (Hz)')
    plt.ylabel('Potencia (μV²/Hz)')
    plt.title('PSD promedio por centro (interpolado y escalas corregidas)')
    plt.legend()
    plt.yscale('log')
    plt.grid(True)
    plt.tight_layout()
    plt.show()

raw_dict_corregido = corregir_escala_a_uV(raw_dict)
plot_psd_all_centers_interpolated(raw_dict_corregido)
```
